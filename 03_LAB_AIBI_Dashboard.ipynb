## LAB 03 - Análise Exploratória com o Dashboard AI/BI

### 🎯 O que vamos aprender neste laboratório?

Neste lab, você vai criar seu primeiro **Dashboard AI/BI** — um painel visual e interativo para analisar dados de forma gráfica, sem precisar escrever código.

> 💡 **O que é um Dashboard?** É uma página com gráficos, tabelas e números que resumem seus dados visualmente. Pense nele como um "painel de controle" — assim como o painel de um carro mostra velocidade, combustível e temperatura de forma rápida, um dashboard mostra indicadores do seu negócio.

> 🤖 **O que é o AI/BI?** É a tecnologia de inteligência artificial do Databricks aplicada ao dashboard. Ela permite que você faça perguntas em português e o sistema gere visualizações automaticamente!

### 📝 Pré-requisitos:
- Ter completado o **Lab 01** (as tabelas devem estar criadas no seu schema)
- Saber o nome do seu `catalog` e `schema` (ex: `dev_treinamento_databricks.maria_silva`)

---
### 🛠️ Passo 1 - Acessando a tabela pelo Catálogo

Vamos começar navegando até a tabela de dados pelo **Catalog Explorer** (Explorador de Catálogo). É a partir dela que vamos criar o dashboard.

| Passo | O que fazer |
|-------|------------|
| 1️⃣ | No menu lateral esquerdo, clique em **"Catalog"** (ícone de banco de dados) |
| 2️⃣ | Navegue pela árvore de catálogos: **prd_treinamento_databricks** → **nome_sobrenome** → **faturamento** |
| 3️⃣ | Você está na página da tabela! Aqui você pode ver as colunas, dados de amostra e metadados |

> 💡 **O que é o Catalog Explorer?** É a interface visual para navegar por todos os dados disponíveis no Databricks. Funciona como um explorador de arquivos, mas para tabelas de dados.

> ⚠️ **Lembre-se:** Seu schema identificado como: `nome_sobrenome` foi definido no Lab 01 (ex: `maria_silva`)

### 🔗 Passo 2 - Criando o Dashboard a partir da tabela

Agora que estamos na página da tabela `faturamento`, vamos criar o dashboard diretamente a partir dela. Isso já conecta os dados automaticamente!

| Passo | O que fazer |
|-------|------------|
| 1️⃣ | Na página da tabela, clique no botão **"Create"** (ou "Criar") no canto superior direito |
| 2️⃣ | Selecione a opção **"Dashboard"** |
| 3️⃣ | Um novo dashboard será criado com a tabela `faturamento` já conectada como dataset! |
| 4️⃣ | Clique no título ("New Dashboard") e renomeie para: **"Lab 03 - Análise Faturamento - [seu nome]"** |

> 🎉 **Vantagem desse método:** A tabela já vem configurada como dataset automaticamente — você não precisa adicionar manualmente!

> ℹ️ **Entendendo a interface do Dashboard:** O dashboard tem duas áreas principais:
> - **Canvas** (centro) → Onde ficam os gráficos e tabelas
> - **Data** (painel lateral) → Onde ficam os datasets (fontes de dados). Você verá que `faturamento` já aparece aqui!

---

> 💡 **O que é um Dataset?** É a "fonte de dados" do dashboard. Cada dataset representa uma tabela ou consulta SQL. Você pode ter vários datasets em um mesmo dashboard.

### Conectando o Dashboard ao SQL Warehouse

Para que o dashboard consiga acessar e processar os dados, ele precisa estar conectado a um **SQL Warehouse**. O warehouse utilizado neste treinamento é:

> **`cluster_sql_treinamento_dtbks`**

#### Como conectar:

| Passo | O que fazer |
|-------|------------|
| 1️⃣ | No canto superior direito do dashboard, localize a bolinha verde que indica se a warehouse está ligada (verde total) ou não conectada (verde contornado) |
| 2️⃣ | Clique nela e uma lista de warehouses disponíveis vai aparecer |
| 3️⃣ | Selecione a warehouse de SQL **`cluster_sql_treinamento_dtbks`**.|
| 4️⃣ | Aguarde até o status mudar para **conectado** (indicador verde) |
---

#### 🤔 Qual a diferença entre o cluster dos labs anteriores e este SQL Warehouse?

Nos Labs 01 e 02, usamos um **cluster All-Purpose** (uso geral). Agora, no dashboard, usamos um **SQL Warehouse**. Veja a diferença:

| Característica | Cluster All-Purpose | SQL Warehouse |
|---------------|--------------------|--------------|
| **Para que serve** | Executar código Python, SQL, Scala — notebooks em geral | Executar **apenas SQL** — otimizado para consultas e dashboards |
| **Usado em** | Notebooks, experimentação, engenharia de dados | Dashboards, consultas SQL, relatórios |
| **Performance SQL** | Boa | **Excelente** (otimizado para isso) |
| **Custo** | Mais caro (roda o tempo todo enquanto ativo) | Mais econômico (escala automaticamente e desliga quando não usado) |
| **Nosso cluster** | `cluster_dbw_treinamento_dtbks` | `cluster_sql_treinamento_dtbks` |

> 💡 **Resumindo:** O SQL Warehouse é como um motor especializado em consultas SQL — mais rápido e econômico para dashboards. Já o cluster All-Purpose é um motor versátil que aceita várias linguagens, ideal para notebooks.

### 🔍 Passo 4 - Adicionando filtros interativos

Filtros permitem que quem visualiza o dashboard **interaja** com os dados, escolhendo o que quer ver (ex: filtrar por bairro, por faixa de atraso, etc.).

| Passo | O que fazer |
|-------|------------|
| 1️⃣ | No menu de criação, presente na parte inferior do canvas, selecione o ícone de fúnil/filtro **"Add a filter"** |
| 2️⃣ | Posicione onde desejar |
| 3️⃣ | Escolha o tipo de filtro. Vamos usar **multiple values**|
| 4️⃣ | Escolha a **coluna** que deseja filtrar (ex: genero_cliente e bairro) no campo **"Fields"** do menu direito |

> 🎯 **Experimente:** Adicione um filtro por "bairro" e outro por "genero_cliente" e veja como **todos os gráficos** se atualizam automaticamente quando você seleciona um valor!

> 💡 **Por que filtros são importantes?** Eles transformam um dashboard estático em uma ferramenta de exploração. Qualquer pessoa pode responder suas próprias perguntas sem precisar editar código.

### 🚀 Passo 7 - Publicando e compartilhando

Seu dashboard está pronto! Agora vamos **publicá-lo** para que outras pessoas possam visualizá-lo:

| Passo | O que fazer |
|-------|------------|
| 1️⃣ | Clique no botão **"Publish"** (Publicar) no canto superior direito |
| 2️⃣ | Selecione o modelo de permissão. Vamos usar "Shared Data Permission" |
| 3️⃣ | Pronto! O dashboard agora tem uma versão publicada |

#### Qual a diferença entre Rascunho e Publicado?
| Modo | Quem vê | Para que serve |
|------|---------|---------------|
| **Draft** (Rascunho) | Só você ou quem tiver permissão de gerenciar/editar | Editar, testar, fazer ajustes |
| **Published** (Publicado) | Todos com acesso | Versão final para visualização |

#### 🔐 Modelos de permissão de dados:

Ao publicar, você deve escolher como os dados serão acessados por quem visualiza o dashboard:

| Modelo | Como funciona | Quando usar |
|--------|--------------|-------------|
| **Shared data permission** | Todos que acessam o dashboard veem **os mesmos dados**, usando as credenciais de quem publicou | Quando você quer garantir que todos vejam as mesmas informações, independente das permissões individuais |
| **Individual data permission** | Cada pessoa vê apenas **os dados que ela tem permissão** de acessar, usando suas próprias credenciais | Quando existem restrições de acesso e cada usuário só deve ver o que lhe é permitido |

> 💡 **Para este laboratório**, selecione **Shared data permission** — assim todos os participantes conseguirão visualizar o dashboard mesmo sem acesso direto às tabelas.

> 💡 Você pode continuar editando o rascunho depois de publicar. As alterações só aparecem para os outros quando você publicar novamente.

| 5️⃣ | Revise o resultado e ajuste o que for necessário |

> 💡 **Quando usar cada abordagem?**
> | Abordagem | Melhor para |
> |-----------|-------------|
> | **Manual (Passo 3)** | Criar visualizações específicas com controle total |
> | **Genie Code lateral** | Gerar um dashboard completo rapidamente, ideal para exploração inicial |

> ⚠️ **Importante:** Mesmo usando o Genie Code para gerar tudo automaticamente, sempre revise os gráficos gerados! Verifique se os dados e agrupamentos fazem sentido para a sua análise.
