### 👇 Célula de trabalho (SQL)

A célula abaixo contém um comando SQL básico que mostra todos os dados da tabela `faturamento`. É nesta célula que o **Genie Code vai inserir o código** gerado a partir das suas perguntas.

**O que significa `SELECT * FROM tabela`?**
- `SELECT` = "me mostre"
- `*` = "todas as colunas"
- `FROM tabela` = "desta tabela"

Ou seja: "Me mostre todas as colunas da tabela faturamento" 😉

> ⚠️ **Antes de usar o Genie Code:** Execute esta célula primeiro (clique no ▶️ play da célula) para confirmar que a tabela existe e os dados estão acessíveis. Se der erro, volte e verifique se o `schema_name` está correto.

```python
%sql
-- 👇 Esta é a célula onde o Genie Code vai gerar código para você!
-- Execute primeiro para ver os dados, depois use Ctrl+I ou o ícone da lâmpada para fazer perguntas.

SELECT * FROM dev_treinamento_databricks.<seu_nome_sobrenome>.faturamento
```
